### Problem Statement:

Kavya is working on a text formatting feature for a messaging application that analyzes user 
readability. One of the checks she needs to implement is to find the longest continuous part 
of a message where no tw e adjacent characters are the same. Her task is to write a program to 
determine the length of the longest substring where adjacent characters (check current and 
previous) are always different.

Input Format:

A string inStr that contains only lowercase English characters

Example 1 :
Input :"xyxyzzz"

Output :5 

Example 2:
Input :"vvuvuvuvuu"
Output:8

```python
substr = input("Enter the string:") # we are giving to users to give string

count = 1      # Counts the current Vaild string
maxcount = 1   

for i in range(1, len(substr)):
    if substr[i] != substr[i - 1]:
        count = count + 1
    else:
        count = 1 
    maxcount = max(maxcount, count)

print(maxcount)
```

### Problem Statement

Food Corner home delivers vegetarian and non-vegetarian meals to its customers based on the order.
A vegetarian combo costs $12 per plate and a non-vegetarian combo costs $15 per plate. 
Apart from the cost per plate of food, customers are also charged for home delivery based on 
the distance in kms from the restaurant to the delivery point. The delivery charges are as 
mentioned

Distance--

Delivery charge per km
First 3km
$0
Next 3km
$1
Remaining kms
$2

Given the type of food, quantity (no. of plates) and the distance in kms from the restaurant 
to the delivery point, implement a program to calculate the final bill amount to be paid by a 
customer.

The below information must be used to check the validity of the data provided by the customer.
Type of food must be 'V' for vegetarian and 'N' for non-vegetarian.
Distance in kms must be greater than 0.
Quantity ordered should be minimum 1.
If any of the input is invalid, the bill amount should be displayed as -1.

Sample Input:                                   | Expected Output:
Food Type     Quantity Ordered   Distance (kms) |
   N                 2               3          |     30
   V                 1               7          |     17

```python
food = input("Enter food type (V/N): ")
quantity = int(input("Enter quantity: "))
distance = float(input("Enter a distance: "))

if (food != "V" and food != "N") or quantity < 1 or distance <= 0:
    bill = -1
else:
    if food == "V":
        price = 12
    else:
        price = 15
    Food_cost = price * quantity

    if distance <= 3:
        delivery = 0
    elif distance <= 6:
        delivery = (distance - 3) * 1 
    else:
        delivery = 3 + (distance - 6) * 2
    bill = Food_cost + delivery 

print(bill)
```

### Problem Statement

Implement a program to calculate the product of three positive integer values. 
However, if one of the integers is 7, consider only the values to the right of 7 for calculation. 
If 7 is the last integer, then display -1.

Note: Only one of the three values can be 7.

Sample Input : 1,5,3; 3,7,8; 7,2,9.

Expected Output :15; 8; 18.

```python
a = int(input("Enter a Num1: "))
b = int(input("Enter a Num2: "))
c = int(input("Enter a Num3: "))

if a == 7:
    result = b * c
elif b == 7:
    result = c
elif c == 7:
    result = -1
else:
    result = a * b * c

print(result)
```
