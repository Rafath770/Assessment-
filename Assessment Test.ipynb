### Problem Statement:

Kavya is working on a text formatting feature for a messaging application that analyzes user 
readability. One of the checks she needs to implement is to find the longest continuous part 
of a message where no tw e adjacent characters are the same. Her task is to write a program to 
determine the length of the longest substring where adjacent characters (check current and 
previous) are always different.

Input Format:

A string inStr that contains only lowercase English characters

Example 1 :
Input :"xyxyzzz"

Output :5 

Example 2:
Input :"vvuvuvuvuu"
Output:8

```python
substr = input("Enter the string:") # we are giving to users to give string

count = 1      # Counts the current Vaild string
maxcount = 1   

for i in range(1, len(substr)):
    if substr[i] != substr[i - 1]:
        count = count + 1
    else:
        count = 1 
    maxcount = max(maxcount, count)

print(maxcount)
```
